# ***Track C : Personal Habit Tracker***

The program is build to track your daily habits for self improvement. You can log in for what you did each day. The tracker will then provide you with your streak (habits performed every consecutive day) and a combined weekly completion rate.

How to run it — Run all cells, then call main() to start.


```python
name = input("Enter your name: ")
print(f"Hello,", name,"!", "Welcome to Habit Loop!")
```

```python
def add_habits():
    habits = []
    while True:
        habit = input("Which habit do you wish to track? ")
        if habit.lower() == "done":
            break
        habits.append(habit)
    return habits

my_habits = add_habits()
print(my_habits)
```

```python
def log_today(my_habits):
  for habit in my_habits:
    habit = input(f"Completed {habit}? Yes or No: ")
log_today(my_habits)
```
