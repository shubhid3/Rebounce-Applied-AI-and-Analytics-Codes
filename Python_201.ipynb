# ***Track C : Personal Habit Tracker***

The program is build to track your daily habits for self improvement. You can log in for what you did each day. The tracker will then provide you with your streak (habits performed every consecutive day) and a combined weekly completion rate.

How to run it — Run all cells, then call main() to start.


```python
def welcome():
  name = input("Enter your name: ")
  print(f"Hello,", name,"!", "Welcome to Habit Loop!")

welcome()
```

```python
def add_habits():
    habits = []
    while True:
        habit = input("Which habit do you wish to track? ")
        if habit.lower() == "done":
            break
        habits.append(habit)
    return habits

my_habits = add_habits()
print(my_habits)
```

```python
Log_habits = {}
from datetime import date
Date = date.today()
def logged_today(my_habits):
  for habit in my_habits:
    answer = input(f"Completed {habit}? Yes or No: ")
    if answer.lower() == "yes":
      if habit not in Log_habits:
        Log_habits[habit] = []
      Log_habits[habit].append(Date)
      print("Yay, Keep it up!")
    else:
      if habit not in Log_habits:
        Log_habits[habit] = []
      print("Oops, Let's do better tomorrow!")


logged_today(my_habits)
print(Log_habits)
```

```python
def currentstreak(habit):
  from datetime import date, timedelta
  Date = date.today()
  current_day = Date
  streak = 0
  while True:
    if current_day in Log_habits[habit]:
      streak = streak + 1
      current_day = current_day - timedelta(days=1)
      print(streak)
    else:
      break
    return streak
print (streak)
for habit in my_habits:
    streak = currentstreak(habit)
    print(f"{habit}: {streak} day streak")
```
