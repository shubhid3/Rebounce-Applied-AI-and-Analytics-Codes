# ***Track C : Personal Habit Tracker***

The program is build to track your daily habits for self improvement. You can log the habit you wish to track. Each day, you can log whether you completed the habit or not. The tracker will then provide you with your streak (habits performed every consecutive day) and a combined weekly completion rate.

How to run it: Run all cells, then call main() to start.


### ***Welcoming the user to the program***

```python
def welcome():
  name = input("Enter your name: ")
  print(f"Hello,", name,"!", "Welcome to Habit Loop!🎯")
```

### ***Defining the function add_habits for the user to add the habit they wish to track.***

```python
def add_habits():
  habits = []
  while True:
    habit = input("Which habit do you wish to track? ")
    if habit.lower() == "done":
      break               #let's the user exit the input function after "done"
    habits.append(habit)
  return habits

my_habits = add_habits()
print(my_habits)
```

### ***Defining Log_habits for the user to track the habits and the dates on which they completed the habits***

```python
Log_habits = {}
from datetime import date                       #imported date from datetime
Date = date.today()
def logged_today(my_habits):
  for habit in my_habits:
    answer = input(f"Completed {habit}? Yes or No: ")
    if answer.lower() == "yes":
      if habit not in Log_habits:
        Log_habits[habit] = []
      Log_habits[habit].append(Date)
      print("Yay, Keep it up!🎉")
    else:
      if habit not in Log_habits:
        Log_habits[habit] = []
      print("Oops, Let's do better tomorrow!💪")


logged_today(my_habits)
print(Log_habits)
```

### ***Defining Current streak for each habit for the user to have a record of the number of days they completed each habit.***

```python
def currentstreak(habit):
  from datetime import date, timedelta
  Date = date.today()
  current_day = Date
  streak = 0
  while True:
    if current_day in Log_habits[habit]:
      streak = streak + 1
      current_day = current_day - timedelta(days=1)       # move back one day to check next
    else:
      break
  return streak

for habit in my_habits:
  streak = currentstreak(habit)
  if streak == 0 and habit in Log_habits and len(Log_habits[habit]) > 0:
    recent_date = max(Log_habits[habit])
    gap_days = (date.today() - recent_date).days
    print(f"{habit}: 0-day streak (last done {gap_days} days ago)")  #counting gap days
  else:
    print(f"{habit}: {streak}-day streak🔥")
```

### ***Defining weekly completing rate in the form of percentage for the user to record each habit.***

```python
def weekly_completion_rate():
  from datetime import date, timedelta
  Date = date.today()
  for habit in my_habits:
    completed_days = 0
    for i in range(7):
      check_day = Date - timedelta(days=i)      #checks completion of habit for all 7 days to provide weekly completion rate
      if check_day in Log_habits[habit]:
        completed_days = completed_days + 1
    percentage = completed_days/7 * 100
    print(f"{habit}: {percentage:.2f}%📊")

weekly_completion_rate()


```

### ***Importing json and saving data to a json file***

```python
import json
def save_data():
  with open("habits.json", "w") as file:
    data = {"habits": my_habits, "logs": {}}
    for habit in Log_habits:
      data["logs"][habit] = []
      for logged_date in Log_habits[habit]:
        data["logs"][habit].append(str(logged_date))
    json.dump(data, file)
save_data()
```

### ***Loading json file back to the program***

```python
def load_data():
  global my_habits, Log_habits
  if os.path.exists("habits.json"):
    try:
      with open("habits.json", "r") as file:
        data = json.load(file)
        my_habits = data["habits"]
        Log_habits = {}
        for habit in data["logs"]:
          Log_habits[habit] = []
          for logged_date in data["logs"][habit]:
            Log_habits[habit].append(date.fromisoformat(logged_date))
    except json.JSONDecodeError:
      print("Save file was empty or corrupted. Starting fresh.")
      my_habits = []
      Log_habits = {}
  else:
    my_habits = []
    Log_habits = {}
```

```python
load_data()
print(my_habits)
print(Log_habits)
```

### ***Defining main() function to start the program***

```python
def main():
  global my_habits
  welcome()
  print("\n==== HABIT TRACKER ====")
  load_data()
  while True:
    print("\n1. Add habit")
    print("2. Log today")
    print("3. View streaks")
    print("4. View weekly completion rate")
    print("5. Save and quit")

    choice = input("Choose an option:")
    if choice == "1":
      my_habits.extend(add_habits())
    elif choice == "2":
      logged_today(my_habits)
    elif choice == "3":
      for habit in my_habits:
        streak = currentstreak(habit)
        print(f"{habit}: {streak}-day streak")
    elif choice == "4":
      weekly_completion_rate()
    elif choice == "5":
      save_data()
      print("Data saved. Goodbye!👋")
      break
    else:
      print("Invalid option. Please choose 1-5.")
```

### ***Run all cells and then run the main() function below***

```python
main()
```
